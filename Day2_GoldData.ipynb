```python
import yfinance as yf

gold = yf.download("GC=F", start="2006-01-01")

print(gold.head())
```

```python
print("Number of rows and columns:")
print(gold.shape)

print("\nFirst available date:")
print(gold.index.min())

print("\nLast available date:")
print(gold.index.max())

print("\nMissing values:")
print(gold.isnull().sum())
```

```python
close_prices = gold["Close"].squeeze()

highest_price = close_prices.max()
highest_date = close_prices.idxmax()

lowest_price = close_prices.min()
lowest_date = close_prices.idxmin()

print("Highest closing price:", highest_price)
print("Date of highest price:", highest_date)

print("Lowest closing price:", lowest_price)
print("Date of lowest price:", lowest_date)
```

```python
import matplotlib.pyplot as plt

close_prices = gold["Close"].squeeze()

plt.figure(figsize=(14, 6))

plt.plot(close_prices.index, close_prices)

plt.title("Gold Price Trend (2006–2026)")
plt.xlabel("Year")
plt.ylabel("Gold Price (USD per Troy Ounce)")

plt.grid(True)
plt.show()
```

```python
close_prices = gold["Close"].squeeze()

yearly_average = close_prices.groupby(
    close_prices.index.year
).mean()

print(yearly_average)
```

```python
plt.figure(figsize=(14, 6))

plt.bar(yearly_average.index.astype(str), yearly_average.values)

plt.title("Average Gold Price Per Year (2006–2026)")
plt.xlabel("Year")
plt.ylabel("Average Gold Price (USD per Troy Ounce)")

plt.xticks(rotation=45)
plt.grid(axis="y")
plt.show()
```
