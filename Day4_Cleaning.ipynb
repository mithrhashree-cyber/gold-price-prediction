```python
df_clean = df.dropna(subset=["gold"]).copy()
```

```python
df_clean.shape
```

```python
df_clean.isnull().sum()
```

```python
df_clean[df_clean["nifty50"].isnull()][["Date", "nifty50"]].head(10)
```

```python
df_clean[df_clean["nifty50"].notnull()][["Date", "nifty50"]].head()
```

```python
df_clean["nifty50"] = df_clean["nifty50"].ffill()
```

```python
df_clean["nifty50"].isnull().sum()
```

```python
df_clean.isnull().sum()
```

```python
columns_to_fill = [
    "silver",
    "usd_inr",
    "dollar_index",
    "crude_oil",
    "bond_yield",
    "sp500",
    "CPI"
]

df_clean[columns_to_fill] = df_clean[columns_to_fill].ffill()
```

```python
df_clean["indian_gold_10g"] = (
    df_clean["gold"] * df_clean["usd_inr"] / 31.1035 * 10
)
```

```python
df_clean.isnull().sum()
```

```python
df_clean.isnull().sum().sum()
```

```python
df_clean.shape
```

```python
df_clean["gold_previous_day"] = df_clean["gold"].shift(1)
```

```python
df_clean[["Date", "gold", "gold_previous_day"]].head()
```

```python
df_clean["gold_pct_change"] = df_clean["gold"].pct_change() * 100
```

```python
df_clean[["Date", "gold", "gold_pct_change"]].head()
```

```python
df_clean["gold_7day_average"] = (
    df_clean["gold"].rolling(window=7).mean()
)
```

```python
df_clean[["Date", "gold", "gold_7day_average"]].head(10)
```

```python
df_clean[[
    "Date",
    "gold",
    "gold_previous_day",
    "gold_pct_change",
    "gold_7day_average"
]].head(10)
```

```python
df_clean["target"] = df_clean["gold"].shift(-1)
```

```python
df_clean[["Date", "gold", "target"]].head(10)
```

```python
df_clean.isnull().sum()
```

```python
model_df = df_clean.drop(columns=["nifty50"]).dropna().copy()
```

```python
model_df.shape
```

```python
model_df.head()
```

```python
model_df.isnull().sum().sum()
```

```python
model_df.to_csv("gold_cleaned.csv", index=False)
```
