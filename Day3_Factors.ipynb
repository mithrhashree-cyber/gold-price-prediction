```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
```

```python
symbols = {
    "gold": "GC=F",
    "silver": "SI=F",
    "usd_inr": "INR=X",
    "dollar_index": "DX-Y.NYB",
    "crude_oil": "CL=F",
    "bond_yield": "^TNX",
    "nifty50": "^NSEI",
    "sp500": "^GSPC"
}

print(symbols)
```

```python
gold_data = yf.download(
    symbols["gold"],
    start="2006-01-01"
)

print(gold_data.head())
```

```python
print(gold_data.columns)
print(gold_data.shape)
```

```python
gold_close = gold_data["Close"].squeeze()

print(gold_close.head())
print(gold_close.shape)
```

```python
market_data = {"gold": gold_close}

for name, ticker in symbols.items():
    if name != "gold":
        df = yf.download(ticker, start="2006-01-01")
        market_data[name] = df["Close"].squeeze()

print("All market data downloaded!")
print(market_data.keys())
```

```python
for name, series in market_data.items():
    print(name, ":", len(series), "records")
```

```python
market_df = pd.concat(market_data, axis=1)

print(market_df.head())
print("Dataset shape:", market_df.shape)
```

```python
cpi_url = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=CPIAUCSL"
fed_url = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=FEDFUNDS"

cpi_data = pd.read_csv(cpi_url)
fed_data = pd.read_csv(fed_url)

print("CPI DATA")
print(cpi_data.head())

print("FEDERAL FUNDS DATA")
print(fed_data.head())
```

```python
print("CPI columns:", cpi_data.columns)
print("FED columns:", fed_data.columns)

print("\nCPI shape:", cpi_data.shape)
print("FED shape:", fed_data.shape)
```

```python
# Prepare CPI data
cpi_data.columns = ["Date", "CPI"]
cpi_data["Date"] = pd.to_datetime(cpi_data["Date"])
cpi_data["CPI"] = pd.to_numeric(cpi_data["CPI"], errors="coerce")
cpi_data = cpi_data.set_index("Date")

# Prepare Federal Funds Rate data
fed_data.columns = ["Date", "FEDFUNDS"]
fed_data["Date"] = pd.to_datetime(fed_data["Date"])
fed_data["FEDFUNDS"] = pd.to_numeric(fed_data["FEDFUNDS"], errors="coerce")
fed_data = fed_data.set_index("Date")

print("CPI prepared:")
print(cpi_data.head())

print("\nFEDFUNDS prepared:")
print(fed_data.head())
```

```python
# Convert market data index into a Date column
market_daily = market_df.reset_index()

# Sort all datasets by date
market_daily = market_daily.sort_values("Date")
cpi_reset = cpi_data.reset_index().sort_values("Date")
fed_reset = fed_data.reset_index().sort_values("Date")

# Merge CPI with market data
combined = pd.merge_asof(
    market_daily,
    cpi_reset,
    on="Date",
    direction="backward"
)

# Merge Federal Funds Rate
combined = pd.merge_asof(
    combined,
    fed_reset,
    on="Date",
    direction="backward"
)

print(combined.head())
print("Combined dataset shape:", combined.shape)
```

```python
combined["indian_gold_10g"] = (
    combined["gold"] * combined["usd_inr"] / 31.1035 * 10
)

print(combined[["gold", "usd_inr", "indian_gold_10g"]].head())
```

```python
print("Missing values in each column:")
print(combined.isnull().sum())

print("\nDataset shape:")
print(combined.shape)
```

```python
combined.to_csv("gold_factors_raw.csv", index=False)

print("Dataset saved successfully!")
```
